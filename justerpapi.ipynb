```python
import requests
import os
import json
import urllib

AUTHOR_ID = "qHFA5z4AAAAJ"


def serp(endpoint, params):
    url = f"https://api.justserpapi.com/api/v1/google/scholar/{endpoint}"
    headers = {"X-API-Key": os.environ["JUSTSERPAPI_KEY"]}
    response = requests.get(url, headers=headers, params=params)
    print(f"status = {response.status_code}; {endpoint}({params})")
    return response.json()
```

```python
author = serp("author", {"author_id": AUTHOR_ID, "results": 100, "language": "en"})
```

```python
paper_list = [
    dict(
        {
            "citation_id": article["citation_id"],
            "year": int(article["year"]),
            "links": dict(
                {
                    "scholar": f"https://scholar.google.ru/citations?view_op=view_citation&hl=en&user={AUTHOR_ID}&citation_for_view={article["citation_id"]}"
                }
            ),
        }
    )
    for article in author["data"]["articles"]
]
paper_list
```

```python
for paper in paper_list:
    citation = serp(
        "author",
        {
            "author_id": AUTHOR_ID,
            "language": "en",
            "view_op": "VIEW_CITATION",
            "citation_id": paper["citation_id"],
        },
    )["data"]["citations"]
    paper["clusters"] = ",".join(
        [
            scholar_article["versions"]["cluster_id"]
            for scholar_article in citation["scholar_articles"]
            if "versions" in scholar_article.keys()
        ]
    )

    paper["authors"] = citation["authors"]
    paper["title"] = citation["title"]

    if "journal" in citation.keys():
        paper["venue"] = citation["journal"]
        if paper["venue"].lower().startswith("arxiv"):
            paper["type"] = "preprint"
        else:
            paper["type"] = "publication"
    elif "conference" in citation.keys():
        paper["venue"] = citation["conference"]
        paper["type"] = "publication"
    elif "book" in citation.keys():
        paper["venue"] = citation["book"]
        paper["type"] = "publication"
    elif "institution" in citation.keys():
        paper["venue"] = "institution"
        paper["type"] = "thesis"
    else:
        raise KeyError("unknown paper type")


with open("papers.json", "w") as file:
    json.dump(paper_list, file, indent=4)
```
